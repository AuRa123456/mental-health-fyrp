```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =========================================================
# dataset paths
# =========================================================

text_dataset_path = "../data/text_features_phasewise.csv"
audio_dataset_path = "../data/audio_features_phasewise.csv"
multimodal_dataset_path = "../data/multimodal_audio_text_features.csv"

# =========================================================
# load datasets
# =========================================================

text_df = pd.read_csv(text_dataset_path)
audio_df = pd.read_csv(audio_dataset_path)
multimodal_df = pd.read_csv(multimodal_dataset_path)

# =========================================================
# target column name
# change this if your target column is different
# =========================================================

target_column = "label"

# =========================================================
# seaborn style
# =========================================================

sns.set_style("whitegrid")
sns.set_context("talk")

# =========================================================
# function to plot dataset distribution
# =========================================================

def plot_distribution(df, title):

    # count depressed and non-depressed samples
    class_counts = df[target_column].value_counts()

    # create dataframe for plotting
    plot_df = pd.DataFrame({
        "Class": class_counts.index,
        "Count": class_counts.values
    })

    # create figure
    plt.figure(figsize=(8, 6))

    sns.barplot(
        data=plot_df,
        x="Class",
        y="Count"
    )

    plt.title(
        f"Dataset Distribution Graph",
        fontsize=20,
        fontweight='bold'
    )

    plt.xlabel("class", fontsize=14)
    plt.ylabel("number of samples", fontsize=14)

    # show values on top of bars
    for index, value in enumerate(plot_df["Count"]):
        plt.text(
            index,
            value + 1,
            str(value),
            ha='center',
            fontsize=12
        )

    plt.tight_layout()
    plt.show()

# =========================================================
# plot graphs individually
# =========================================================

plot_distribution(
    text_df,
    "Text-Only Model"
)

plot_distribution(
    audio_df,
    "Audio-Only Model"
)

plot_distribution(
    multimodal_df,
    "Multimodal Model"
)

print("all dataset distribution graphs generated successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =========================================================
# multimodal random forest metrics across phases
# =========================================================

data = {
    "Phase": [
        "Phase 1",
        "Phase 2",
        "Phase 3"
    ],

    # random forest 10-fold accuracy values
    "Accuracy": [
        83.60,
        79.29,
        77.85
    ],

    # f1-score multiplied by 100
    "F1-Score": [
        70.97,
        60.00,
        56.76
    ]
}

# =========================================================
# create dataframe
# =========================================================

df = pd.DataFrame(data)

# =========================================================
# convert dataframe to long format
# =========================================================

df_melted = df.melt(
    id_vars="Phase",
    value_vars=["Accuracy", "F1-Score"],
    var_name="Metric",
    value_name="Score"
)

# =========================================================
# seaborn style
# =========================================================

sns.set_style("whitegrid")
sns.set_context("talk")

# =========================================================
# bar graph
# =========================================================

plt.figure(figsize=(12, 7))

ax = sns.barplot(
    data=df_melted,
    x="Phase",
    y="Score",
    hue="Metric"
)

# =========================================================
# graph formatting
# =========================================================

plt.title(
    "Multimodal Random Forest Accuracy and F1 Across Conversational Phases",
    fontsize=22,
    fontweight='bold'
)

plt.xlabel(
    "Conversational Phases",
    fontsize=16
)

plt.ylabel(
    "Score (%)",
    fontsize=16
)

plt.ylim(0, 100)

plt.legend(
    title="Metric",
    fontsize=12,
    title_fontsize=14,
    loc='upper right',
    frameon=True
)

# =========================================================
# add value labels
# =========================================================

for container in ax.containers:

    ax.bar_label(
        container,
        fmt='%.2f',
        fontsize=11
    )

plt.tight_layout()

plt.show()

print("bar graph generated successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =========================================================
# phase 1 multimodal model accuracies
# =========================================================

data = {
    "Model": [
        "Random Forest",
        "Random Forest 5 Folds",
        "Random Forest 10 Folds",
        "XGBoost",
        "XGBoost 5 Folds",
        "XGBoost 10 Folds",
        "SVM",
        "SVM 5 Folds",
        "SVM 10 Folds",
        "Logistic Regression",
        "Logistic Regression 5 Folds",
        "Logistic Regression 10 Folds",
        "CatBoost",
        "CatBoost 5 Folds",
        "CatBoost 10 Folds",
        "FeedForward Neural Network"
    ],

    "Accuracy": [
        83.89,
        82.76,
        83.60,
        65.56,
        81.75,
        82.32,
        78.33,
        80.54,
        82.49,
        78.33,
        82.54,
        82.97,
        74.44,
        80.64,
        80.11,
        80.56
    ]
}

# =========================================================
# create dataframe
# =========================================================

df = pd.DataFrame(data)

# =========================================================
# sort models by accuracy
# =========================================================

df = df.sort_values(
    by="Accuracy",
    ascending=False
)

# =========================================================
# seaborn style
# =========================================================

sns.set_style("whitegrid")
sns.set_context("talk")

# =========================================================
# phase 1 accuracy comparison bar graph
# =========================================================

plt.figure(figsize=(20, 9))

ax = sns.barplot(
    data=df,
    x="Model",
    y="Accuracy"
)

# =========================================================
# graph formatting
# =========================================================

plt.title(
    "Phase 1 Accuracy Comparison Across All Multimodal Models",
    fontsize=24,
    fontweight='bold'
)

plt.xlabel(
    "Models",
    fontsize=16
)

plt.ylabel(
    "Accuracy (%)",
    fontsize=16
)

plt.ylim(0, 100)

plt.xticks(
    rotation=65,
    ha='right'
)

# =========================================================
# add accuracy labels
# =========================================================

for container in ax.containers:

    ax.bar_label(
        container,
        fmt='%.2f',
        fontsize=10
    )

plt.tight_layout()

plt.show()

print("phase 1 multimodal accuracy comparison graph generated successfully!")
```
