Text Only

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# data
data = {
    "Models": [
        "Random Forest",
        "Random Forest 5 Folds",
        "Random Forest 10 Folds",
        "XGBoost",
        "XGBoost 5 Folds",
        "XGBoost 10 Folds",
        "SVM",
        "SVM 5 Folds",
        "SVM 10 Folds",
        "Logistic Regression",
        "Logistic Regression 5 Folds",
        "Logistic Regression 10 Folds",
        "CatBoost",
        "CatBoost 5 Folds",
        "CatBoost 10 Folds"
    ],

    "Phase 1": [
        60.83, 71.43, 74.54,
        73.89, 64.96, 68.59,
        61.11, 58.42, 54.35,
        55.56, 58.90, 61.78,
        78.33, 80.76, 81.75
    ],

    "Phase 2": [
        71.30, 68.42, 69.85,
        58.33, 63.01, 64.94,
        42.78, 48.05, 43.75,
        45.00, 50.05, 48.80,
        72.22, 75.04, 78.60
    ],

    "Phase 3": [
        53.06, 61.32, 57.71,
        52.78, 60.41, 60.85,
        54.44, 48.10, 51.13,
        43.89, 46.15, 46.49,
        71.11, 67.22, 65.15
    ]
}

# dataframe
df = pd.DataFrame(data)

# Melt dataframe for seaborn
df_melted = df.melt(
    id_vars="Models",
    var_name="Phase",
    value_name="Accuracy"
)

# seaborn style
sns.set_style("whitegrid")
sns.set_context("talk")

# 1. grouped bar chart
plt.figure(figsize=(20, 9))

sns.barplot(
    data=df_melted,
    x="Models",
    y="Accuracy",
    hue="Phase"
)

plt.ylim(0, 100)

plt.title("Text-Only Model Performance Comparison Across Phases",
          fontsize=20,
          fontweight='bold')

plt.xlabel("Models", fontsize=14)
plt.ylabel("Accuracy (%)", fontsize=14)

plt.xticks(rotation=75)

plt.legend(title="Phases")

plt.tight_layout()
plt.show()

# 2. heatmap
plt.figure(figsize=(12, 9))

heatmap_data = df.set_index("Models")

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".2f",
    cmap="YlGnBu",
    linewidths=0.5,
    vmin=0,
    vmax=100
)

plt.title("Text-Only Model Accuracy Heatmap Across Phases",
          fontsize=20,
          fontweight='bold')

plt.tight_layout()
plt.show()

# 3. line plot
plt.figure(figsize=(20, 9))

for phase in ["Phase 1", "Phase 2", "Phase 3"]:

    sns.lineplot(
        x=df["Models"],
        y=df[phase],
        marker='o',
        linewidth=3,
        markersize=10,
        label=phase
    )

plt.ylim(0, 100)

plt.title("Text-Only Model Phase-wise Accuracy Trends",
          fontsize=20,
          fontweight='bold')

plt.xlabel("Models", fontsize=14)
plt.ylabel("Accuracy (%)", fontsize=14)

plt.xticks(rotation=75)

plt.legend()

plt.tight_layout()
plt.show()

# 4. boxplot
plt.figure(figsize=(10, 7))

sns.boxplot(
    data=df_melted,
    x="Phase",
    y="Accuracy"
)

plt.ylim(0, 100)

plt.title("Text-Only Model Accuracy Distribution Across Phases",
          fontsize=20,
          fontweight='bold')

plt.xlabel("Phases", fontsize=14)
plt.ylabel("Accuracy (%)", fontsize=14)

plt.tight_layout()
plt.show()

# 6. top models
top_models = df[
    df["Models"].isin([
        "Random Forest 10 Folds",
        "XGBoost",
        "CatBoost",
        "CatBoost 5 Folds",
        "CatBoost 10 Folds"
    ])
]

top_melted = top_models.melt(
    id_vars="Models",
    var_name="Phase",
    value_name="Accuracy"
)

# Bigger figure size
plt.figure(figsize=(18, 10))

sns.barplot(
    data=top_melted,
    x="Phase",
    y="Accuracy",
    hue="Models"
)

plt.ylim(0, 100)

plt.title("Top Performing Text-Only Models Comparison",
          fontsize=20,
          fontweight='bold')

plt.xlabel("Phases", fontsize=14)
plt.ylabel("Accuracy (%)", fontsize=14)

plt.legend(
    title="Models",
    fontsize=12,
    title_fontsize=14,
    loc='upper right',
    frameon=True
)

plt.tight_layout()
plt.show()

# 7. phase wise separate bar charts
phases = ["Phase 1", "Phase 2", "Phase 3"]

for phase in phases:

    plt.figure(figsize=(18, 8))

    sorted_df = df.sort_values(by=phase, ascending=False)

    sns.barplot(
        data=sorted_df,
        x="Models",
        y=phase
    )

    plt.ylim(0, 100)

    plt.title(f"{phase} Accuracy Comparison for Text-Only Models",
              fontsize=20,
              fontweight='bold')

    plt.xlabel("Models", fontsize=14)
    plt.ylabel("Accuracy (%)", fontsize=14)

    plt.xticks(rotation=75)

    plt.tight_layout()
    plt.show()

print("All Text-Only graphs generated successfully!")
```
