```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =========================================================
# best model accuracy from each modality
# based on highest accuracy values provided above
# =========================================================

comparison_data = {
    "Modality": [
        "Audio-Only",
        "Text-Only",
        "Multimodal"
    ],

    "Best Model": [
        "Random Forest 5 Folds",
        "CatBoost",
        "Random Forest"
    ],

    # highest accuracies from the provided tables
    "Accuracy": [
        67.62,   # audio-only best
        85.56,   # text-only best
        83.89    # multimodal best
    ]
}

# =========================================================
# create dataframe
# =========================================================

df = pd.DataFrame(comparison_data)

# =========================================================
# seaborn style
# =========================================================

sns.set_style("whitegrid")
sns.set_context("talk")

# =========================================================
# audio vs text vs multimodal comparison graph
# =========================================================

plt.figure(figsize=(10, 7))

ax = sns.barplot(
    data=df,
    x="Modality",
    y="Accuracy"
)

plt.title(
    "Audio vs Text vs Multimodal Comparison Graph",
    fontsize=22,
    fontweight='bold'
)

plt.xlabel("modality", fontsize=16)
plt.ylabel("accuracy (%)", fontsize=16)

plt.ylim(0, 100)

# =========================================================
# add accuracy labels on bars
# =========================================================

for i, value in enumerate(df["Accuracy"]):

    ax.text(
        i,
        value + 1,
        f"{value:.2f}%",
        ha='center',
        fontsize=14,
        fontweight='bold'
    )

# =========================================================
# add best model names below bars
# =========================================================

for i, model in enumerate(df["Best Model"]):

    ax.text(
        i,
        5,
        model,
        ha='center',
        fontsize=11,
        rotation=0
    )

plt.tight_layout()
plt.show()

print("audio vs text vs multimodal comparison graph generated successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =========================================================
# best configuration per model
# highest accuracy selected for each model
# multimodal results used
# =========================================================

data = {
    "Model": [
        "Random Forest",
        "XGBoost",
        "SVM",
        "Logistic Regression",
        "CatBoost"
    ],

    "Best Configuration": [
        "10 Folds",
        "10 Folds",
        "10 Folds",
        "10 Folds",
        "5 Folds"
    ],

    "Best Accuracy": [
        83.60,   # random forest 10 folds
        82.32,   # xgboost 10 folds
        82.49,   # svm 10 folds
        82.97,   # logistic regression 10 folds
        80.64    # catboost 5 folds
    ]
}

# =========================================================
# create dataframe
# =========================================================

df = pd.DataFrame(data)

# =========================================================
# seaborn style
# =========================================================

sns.set_style("whitegrid")
sns.set_context("talk")

# =========================================================
# model accuracy comparison bar chart
# =========================================================

plt.figure(figsize=(14, 8))

ax = sns.barplot(
    data=df,
    x="Model",
    y="Best Accuracy"
)

plt.title(
    "Model Accuracy Comparison Bar Chart (Best Config per Model)",
    fontsize=22,
    fontweight='bold'
)

plt.xlabel("models", fontsize=16)
plt.ylabel("accuracy (%)", fontsize=16)

plt.ylim(0, 100)

plt.xticks(fontsize=13)
plt.yticks(fontsize=13)

# =========================================================
# add accuracy labels
# =========================================================

for i, value in enumerate(df["Best Accuracy"]):

    ax.text(
        i,
        value + 1,
        f"{value:.2f}%",
        ha='center',
        fontsize=13,
        fontweight='bold'
    )

# =========================================================
# add configuration labels
# =========================================================

for i, config in enumerate(df["Best Configuration"]):

    ax.text(
        i,
        5,
        config,
        ha='center',
        fontsize=11
    )

plt.tight_layout()

plt.show()

print("model accuracy comparison bar chart generated successfully!")
```
